```python
import pandas as pd
import numpy as np
import yfinance as yf
import os
```

```python
df = pd.read_csv('D:\Project_Github\set_fear_index\dataset\SET50_constituents_2016_2026.csv')
df
```

```python
for i in range(0, len(df)):
    print(df['period'][i], df['symbol'][i])
```

```python
ticker_map = {
    'TMB.BK': 'TTB.BK',
    'MTLS.BK': 'MTC.BK',
    'INTUCH.BK': 'GULF.BK',
    'GLOW.BK': 'GPSC.BK',
    'ROBINS.BK': 'CRC.BK',
    'DTAC.BK': 'TRUE.BK',
    'PS.BK': 'PSH.BK',
}


df['symbol'] = df['symbol'].replace(ticker_map)
```

```python
df['symbol'] = df['symbol']+'.BK'
```

```python
df['symbol'].tolist()
```

```python
set50 = yf.download(
    df['symbol'].tolist(),
    period="10y",
    interval="1d",
    auto_adjust=True,
    group_by="ticker"
)
```

```python
set50
```

```python
output_dir = r"D:\Project_Github\set_fear_index\dataset\parquet"

# ดึงรายชื่อหุ้นจาก MultiIndex
tickers = set50.columns.get_level_values("Ticker").unique()

summary = []

for ticker in tickers:

    # แยกข้อมูลของแต่ละหุ้น
    stock = set50.xs(
        ticker,
        axis=1,
        level="Ticker"
    ).copy()

    # ลบแถวและคอลัมน์ที่ไม่มีข้อมูลทั้งหมด
    stock = stock.dropna(how="all")
    stock = stock.dropna(axis=1, how="all")

    if stock.empty:
        continue

    # จัดการ Index
    stock.index.name = "Date"

    # บันทึก Parquet
    filename = f"{ticker}.parquet"

    stock.to_parquet(
        os.path.join(output_dir, filename),
        engine="pyarrow",
        compression="snappy"
    )

    summary.append({
        "Ticker": ticker,
        "Start_Date": stock.index.min(),
        "End_Date": stock.index.max(),
        "Rows": len(stock),
        "Columns": len(stock.columns)
    })

summary_df = pd.DataFrame(summary)

summary_df.to_parquet(
    os.path.join(output_dir, "stock_summary.parquet"),
    index=False
)

print("Saved:", len(summary_df), "stocks")
```
